```python
# print("\n Calculadora Multiple Calcula o Convierte Valores \n " )
print("\n Elige una opcion \n")
print(" 1 = SUMA  ")
print(" 2 = RESTA  ")
print(" 3 = MULTIPLICACION  ")
print(" 4 = DIVICION  ")
print(" 5 = POTENCIA  ")
print(" 6 = Convertidor de decimal a Binario")
print(" 7 = Convertidor de decimal a Octal")
print(" 8 = Convertidor de decimal a Hexadecimal")


opcion = input (" \n---Digite una opcion:---")

if opcion == '1' :
    print( "\n---SUMA DE 2 NUMEROS---\n")
    num1= float(input("\nDigite un numero:>>"))
    num2= float(input("\nDigite un numero:>>"))
    suma = num1+num2
    print(f" \n El Resultado es {suma}")

elif opcion == '2' :
     print( "\n--RESTA DE 2 NUMEROS--\n")
     num1= float(input("\nDigite un numero:>>"))
     num2= float(input("\nDigite un numero:>>"))
     resta = num1-num2
     print(f" \n El Resultado es {resta}")

elif opcion == '3' :
     print( "\n--MULTIPLICACION DE 2 NUMEROS --\n")
     num1= float(input("\nDigite un numero:>>"))
     num2= float(input("\nDigite un numero:>>"))
     multi = num1*num2
     print(f" \n El Resultado es {multi}")

elif opcion == '4' :
     print( "\n--DIVICION DE 2 NUMEROS--\n")
     num1= float(input("\nDigite un numero:>>"))
     num2= float(input("\nDigite un numero:>>"))
     div = num1/num2
     print(f" \n El Resultado es {div}")

elif opcion == '5' :
     print( "\n--POTENCIA--\n")
     num1= float(input("\nDigite un numero base:>>"))
     num2= float(input("\nDigite un numero potencia:>>"))
     pot = num1-num2
     print(f" \n El Resultado es {pot}")


elif opcion == '6' :
    def binario_decimal():
        print( "---- Convertidor de decimal a binario----")
        numero = int(input("INTRODUCE UN NUMERO DECIMAL ENTERO: \n"))
        convertido = bin (numero)
        print(f" El numero {numero} en binario es :{convertido}")
    binario_decimal()

elif opcion == '7':
        def octal_decimal():
            numero = int(input('INTRODUCE UN NUMERO: \n'))
            convertido = oct(numero)
            print(convertido)
        octal_decimal()

elif opcion == '8':
        def hex_decimal():
            numero = int(input('INTRODUCE UN NUMERO: \n'))
            convertido = hex(numero)
            print(convertido)
        hex_decimal()
    
elif opcion != '1' or '2' or '3' or'4' or '5' or'6' or'7' or '8':
    print("OPCION INCORRECTA")
```
